# Exploratory Analysis of Synthetic EEG

This notebook walks through the main steps of the pipeline interactively.
Run `scripts/run_pipeline.py` for the full automated version.

```python
import sys
from pathlib import Path
ROOT = Path.cwd().parent
sys.path.insert(0, str(ROOT))

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from scripts.generate_synthetic_data import generate_dataset
from preprocessing.filter import apply_standard_pipeline
from analysis.spectral import compute_psd, relative_band_power
from visualization.plots import plot_raw_signal, plot_psd_comparison
```

```python
# Generate a small dataset if it does not exist yet
data_dir = ROOT / "data" / "raw"
if not (data_dir / "synthetic_eeg.npy").exists():
    data, meta = generate_dataset(n_subjects=6, out_dir=data_dir)
else:
    data = np.load(data_dir / "synthetic_eeg.npy")
    meta = pd.read_csv(data_dir / "metadata.csv")

print(data.shape)
meta.head()
```

```python
fs = float(meta["fs"].iloc[0])
example = {}
for state in ["awake", "sleep", "anesthesia"]:
    idx = meta.index[meta["state"] == state][0]
    example[state] = apply_standard_pipeline(data[idx], fs=fs)

plot_psd_comparison(example, fs=fs)
plt.show()
```
